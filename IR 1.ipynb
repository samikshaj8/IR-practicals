```python
import nltk
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text = """
Natural Language Processing is an important field of Artificial Intelligence.
Students are learning different techniques of processing text data.
They are studying, playing, and working on interesting projects.
"""

print("Original Text:")
print(text)
```

```python
text = text.lower()

print("Lowercase Text:")
print(text)
```

```python
# Remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

print("Text After Removing Punctuation:")
print(text)
```

```python
# perform tokenization
tokens = word_tokenize(text)

print("Tokenized Words:")
print(tokens)
```

```python
# Remove stop words
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word not in stop_words:
        filtered_words.append(word)

print("After Stop Word Removal:")
print(filtered_words)
```

```python
# Perform stemming
stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print("After Stemming:")
print(stemmed_words)
```

```python
# Display final preprocessed text
final_text = " ".join(stemmed_words)

print("Original Text:")
print(text)

print("\nFinal Preprocessed Text:")
print(final_text)
```
