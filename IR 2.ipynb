```python
import re
import nltk

from collections import defaultdict
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
nltk.download('stopwords')
```

```python
# Create sample documents
documents = {
    1: "The quick brown fox jumped over the lazy dog.",
    2: "The lazy dog slept in the sun.",
    3: "The fox is running in the forest.",
    4: "Information retrieval is an important part of search engines."
}

print("Documents:")
for doc_id, content in documents.items():
    print("Document", doc_id, ":", content)
```

```python
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
```

```python
# Define the text preprocessing function
def preprocess(text):

    # Convert text into lowercase
    text = text.lower()

    # Extract words using tokenization
    words = re.findall(r'\b[a-z]+\b', text)

    # Remove stop words and perform stemming
    processed_words = []

    for word in words:
        if word not in stop_words:
            stemmed_word = stemmer.stem(word)
            processed_words.append(stemmed_word)

    return processed_words
```

```python
# Create the inverted index
inverted_index = defaultdict(list)

for doc_id, content in documents.items():

    words = preprocess(content)

    for word in set(words):
        inverted_index[word].append(doc_id)

print("Inverted Index:")

for word, doc_ids in sorted(inverted_index.items()):
    print(word, ":", doc_ids)
```

```python
# search and retrieve documents
def search(query):

    query_words = preprocess(query)

    if not query_words:
        print("Please enter a valid search query.")
        return

    result_docs = set()

    for word in query_words:
        if word in inverted_index:
            result_docs.update(inverted_index[word])

    if result_docs:
        print("\nDocuments found:")

        for doc_id in sorted(result_docs):
            print("\nDocument", doc_id, ":")
            print(documents[doc_id])

    else:
        print("No matching documents found.")
```

```python
query = input("Enter your search query: ")

search(query)
```
